#### Problem 2.2

How many ha should the farmer dedicate to each crop and with what
pesticide application rate(s)? How much profit will the farmer expect to
make?


```julia
farm_model = Model(HiGHS.Optimizer) # initialize model object

Y = [2900 3500 5900; 3800 4100 6700; 4400 4200 7900] #yield matrix for [application rate "j", crop type "i"]

r = [0.36 0.27 0.22] #selling price/revenue ($/kg) for soy, wheat, corn respectively

c = [350 280 390] #production costs ($/ha) for soy, wheat, corn respectively

a = 70 #$/kg of pesticide applied

I = 1:3 #("soy", "wheat", "corn")
J = 1:3 #(0, 1, 2) kg/ha 

@variable(farm_model, 0 <= x[j in J, i in I] <= 130) # land allocation


@objective(farm_model, Max, sum(x[j,i]*(Y[j,i]*r[i] - c[i] - a*j) for j in J, i in I)) #total profit

#max available ha
@constraint(farm_model, max_ha, sum(x[j,i] for j in J, i in I)  <= 130)
#max avg pesticide per crop
@constraint(farm_model, soy_pest, sum(1 * x[1,i] for i in I) <= 0.8*130)
@constraint(farm_model, wheat_pest, sum(2 * x[2,i] for i in I) <= 0.7*130)
@constraint(farm_model, corn_pest, sum(3 * x[3,i] for i in I) <= 0.6*130)
#max demand for each crop
@constraint(farm_model, max_soy, sum(Y[j,1] * x[j,1] for j in J) <= 250000)
@constraint(farm_model, max_wheat, sum(Y[j,2] * x[j,2] for j in J) <= 250000)
@constraint(farm_model, max_corn, sum(Y[j,3] * x[j,3] for j in J) <= 250000)

print(farm_model)

optimize!(farm_model)
@show value.(x);
@show objective_value(farm_model);
```
